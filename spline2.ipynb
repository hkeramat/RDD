```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import splprep, splev

# Load the airfoil coordinate data
df = pd.read_csv('./Airfoiltestcase4.csv')

# Extract x and y coordinates directly
x = df['x'].values
y = df['y'].values

# Create a parametric spline representation of the curve
tck, u = splprep([x, y], s=0.005, k=3)

# Generate interpolated points along the spline
unew = np.linspace(0, 1, 500)
x_smooth, y_smooth = splev(unew, tck)

# Create a DataFrame for the smoothed coordinates
df_smooth = pd.DataFrame({'x': x_smooth, 'y': y_smooth})

# Save the smoothed data to CSV
output_path = './Airfoiltestcase4_smooth.csv'
df_smooth.to_csv(output_path, index=False)

# Plot the smoothed spline curve
plt.figure(figsize=(8, 4))
plt.plot(x_smooth, y_smooth)
plt.axis('equal')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Spline-Fitted Airfoil Shape')
plt.show()

print(f"Smoothed data saved to {output_path}")
```
